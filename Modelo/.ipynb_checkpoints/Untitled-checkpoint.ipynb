```python
#Importar librerias
import pandas as pd
from pathlib import Path
```

```python
# 1. Configuración
carpeta = Path("datasets/datasets_cierre_agricola_raw_data/")
archivos = sorted(carpeta.glob("*.csv"))
print(f"Se encontraron {len(archivos)} archivos")

# 2. Función que prueba varias codificaciones (ñ y acentos)
def leer_csv(ruta):
    for enc in ["utf-8", "utf-8-sig", "cp1252", "latin-1"]:
        try:
            temp = pd.read_csv(ruta, encoding=enc, dtype=str, low_memory=False)
            temp["archivo_origen"] = ruta.name   # opcional: de qué archivo viene
            return temp
        except UnicodeDecodeError:
            continue
    raise ValueError(f"No se pudo leer {ruta}")

# 3. Unir todos los archivos
df = pd.concat((leer_csv(a) for a in archivos), ignore_index=True)

# 4. Convertir a número la columna problemática (sin errores)
col = "Volumenproduccion"
df[col] = pd.to_numeric(df[col], errors="coerce")

# 5. Revisar resultado
print(df.shape)
print(df[col].dtype, "|", df[col].isna().sum(), "valores nulos")
df.head()
```

```python
# Unificar nombre de cultivo
df["Nomcultivo"] = df["Nomcultivo"].fillna(df["Nomcultivo Sin Um"])
df = df.drop(columns=["Nomcultivo Sin Um"])

# Filtrar Durango sin sobrescribir el dataset completo
df_durango = df[df["Nomestado"].str.strip().str.lower() == "durango"].copy()

# Nulos solo en columnas clave
df_durango = df_durango.dropna(subset=["Volumenproduccion"])

print(df_durango.shape)
df_durango.head()
```

```python
#Elegimos las columnas que usemos
columnas_modelo = [
    "Anio", "Nommunicipio", "Nomcicloproductivo", "Nommodalidad",
    "Nomcultivo", "Sembrada", "Cosechada", "Siniestrada",
    "Volumenproduccion", "Rendimiento", "Precio", "Valorproduccion"
]

df_modelo = df_durango[columnas_modelo].copy()
```

```python
#Convertimos a numeros las columnas numericas
cols_num = ["Anio", "Sembrada", "Cosechada", "Siniestrada",
            "Volumenproduccion", "Rendimiento", "Precio", "Valorproduccion"]

df_modelo[cols_num] = df_modelo[cols_num].apply(pd.to_numeric, errors="coerce")
df_modelo = df_modelo.dropna()

df_modelo.info()
```

```python
df_modelo.head()
```

```python
df_modelo.to_csv("durango_modelo.csv", index=False, encoding="utf-8-sig")
```
